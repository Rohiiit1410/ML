```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn import metrics
```

```python
iris = load_iris()

X = iris.data
print("Shape of X:", X.shape)
print("Feature names:", iris.feature_names)
```

```python
scaler = StandardScaler()
X = scaler.fit_transform(X)
print("Data normalized.")
```

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

print("Shape after PCA:", X_pca.shape)
print("Explained Variance Ratio:", pca.explained_variance_ratio_)
print("Total Variance Explained:", round(sum(pca.explained_variance_ratio_) * 100, 2), "%")
```

```python
# ── K-Means Clustering ──

kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
kmeans_labels = kmeans.fit_predict(X_pca)

print("K-Means Cluster Labels:", np.unique(kmeans_labels))
print("Inertia (Within-cluster SSE):", round(kmeans.inertia_, 4))
print("Silhouette Score:", round(metrics.silhouette_score(X_pca, kmeans_labels), 4))
```

```python
plt.figure(figsize=(7, 5))
scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=kmeans_labels, cmap='viridis', s=60, edgecolors='k', linewidths=0.4)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
            c='red', marker='X', s=200, label='Centroids', zorder=5)
plt.title('K-Means Clustering (PCA-reduced Iris)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend()
plt.colorbar(scatter, label='Cluster')
plt.tight_layout()
plt.show()
```

```python
# ── Agglomerative Clustering ──

agg = AgglomerativeClustering(n_clusters=3, linkage='ward')
agg_labels = agg.fit_predict(X_pca)

print("Agglomerative Cluster Labels:", np.unique(agg_labels))
print("Silhouette Score:", round(metrics.silhouette_score(X_pca, agg_labels), 4))
```

```python
plt.figure(figsize=(7, 5))
scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=agg_labels, cmap='plasma', s=60, edgecolors='k', linewidths=0.4)
plt.title('Agglomerative Clustering (PCA-reduced Iris)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.colorbar(scatter, label='Cluster')
plt.tight_layout()
plt.show()
```

```python
# ── Dendrogram (bonus - shows how Agglomerative merges clusters) ──

from scipy.cluster.hierarchy import dendrogram, linkage

linked = linkage(X_pca, method='ward')

plt.figure(figsize=(12, 5))
dendrogram(linked, truncate_mode='lastp', p=20, leaf_rotation=45, leaf_font_size=10)
plt.title('Dendrogram - Agglomerative Clustering (Ward Linkage)')
plt.xlabel('Sample Index / Cluster Size')
plt.ylabel('Distance')
plt.tight_layout()
plt.show()
```

```python
# ── Side-by-side Comparison ──

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].scatter(X_pca[:, 0], X_pca[:, 1], c=kmeans_labels, cmap='viridis', s=60, edgecolors='k', linewidths=0.4)
axes[0].scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
                c='red', marker='X', s=200, label='Centroids', zorder=5)
axes[0].set_title('K-Means Clustering')
axes[0].set_xlabel('PC1')
axes[0].set_ylabel('PC2')
axes[0].legend()

axes[1].scatter(X_pca[:, 0], X_pca[:, 1], c=agg_labels, cmap='plasma', s=60, edgecolors='k', linewidths=0.4)
axes[1].set_title('Agglomerative Clustering')
axes[1].set_xlabel('PC1')
axes[1].set_ylabel('PC2')

plt.suptitle('Clustering Comparison on PCA-reduced Iris Dataset', fontsize=13)
plt.tight_layout()
plt.show()

print("\nSummary")
print("K-Means     Silhouette Score:", round(metrics.silhouette_score(X_pca, kmeans_labels), 4))
print("Agglomerative Silhouette Score:", round(metrics.silhouette_score(X_pca, agg_labels), 4))
```
